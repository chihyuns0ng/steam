```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('games.csv')

df.head()
```

```python
df.rename(columns={
    'AppID' : 'Name',
    'Name' : 'Release date',
    'Release date' : 'Estimated owners',
    'Estimated owners' : 'Peak CCU',
    'Peak CCU' : 'Required age',
    'Required age' : 'Price',
    'Price' : 'Discount',
    'DiscountDLC count' : 'DLC Count',
    }, inplace=True)

df = df.reset_index().rename(columns={'index' : 'AppID'})
df.drop(columns=['Movies'], inplace=True)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df['Release date'] = pd.to_datetime(df['Release date'], format='%b %d, %Y', errors='coerce')
df['Release date']
```

```python
df = df[~df['Name'].str.contains('Playtest', na=False)]
df.shape
```

```python
df = df[~df['Name'].isna()]
df.shape
```

```python
df['Full audio languages']
```

```python
df[df['Full audio languages'].astype(str)=='[]']
# 인게임 음성 미지원
```

```python
genre_counts = df['Genres'].str.split(',').explode().value_counts()
print(genre_counts)
```

```python
tag_counts = df['Tags'].str.split(',').explode().value_counts()
print(tag_counts)
```

```python
_ = df['Discount'].hist()
# 분포가 0~100인 것으로 보아 최대 할인율로 추정
```

```python
plt.figure(figsize=(12, 6))

years = df['Release date'].dt.year

sns.countplot(x=years, palette='viridis')

plt.title('Game Releases per Year', fontsize=15)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12, 6))

months = df['Release date'].dt.month

sns.countplot(x=months, palette='viridis')

plt.title('Game Releases per Month', fontsize=15)
plt.xticks(rotation=45)
plt.show()
```
